# Proxy-robust covariance losses

The latent covariance is not observed. Evaluation uses a noisy proxy $S$ with $\mathrm{E}[S\mid\mathcal{F}]=\Sigma$. This notebook uses the production losses in `covharness.losses`. It does not implement a new scoring rule.

Construction. $S = u\Sigma$ with $u\sim\mathrm{Exp}(1)$, so $\mathrm{E}[u]=1$. $H_A=\Sigma$ is the true forecast. $H_B=\log(2)\Sigma$ matches the median of $u$ and is inferior in mean.


```python
from covharness.losses.robustness import (
    analytic_expected_losses,
    monte_carlo_proxy_ranking,
    plot_proxy_robustness,
)

analytic = analytic_expected_losses()
result = monte_carlo_proxy_ranking()
analytic, result
```

Squared Frobenius and reduced QLIKE rank $H_A$ above $H_B$ in expectation. Full Stein agrees when $S$ is SPD. Ordinary unsquared Frobenius ranks $H_B$ first because it tracks the median of the proxy noise rather than the mean. Single-draw flip rates need not match expected-loss rankings.
